```python
# Create a DataFrame by reading the customers table from the specified catalog/database

# Spark session is used to access the table via the DataFrame Reader API
df = (spark      # Initialize the DataFrame using the active SparkSession
      .read      # Use the DataFrame Reader to load data from a source
      .table("main.dbdemos_sdp_cdc.customers"))  # Read the 'customers' table from the main.dbdemos_sdp_cdc schema

# Display the first 10 rows of the DataFrame for a quick preview of the data
df.show(n=10)
```

```python
# Write the customers DataFrame to a managed Unity Catalog table in the workspace.bronze schema

(df                  # Start from the customers DataFrame loaded in the previous cell
 .write              # Access the DataFrameWriter to persist the DataFrame to a target destination
 .mode("overwrite")  # Overwrite the target table if it already exists, replacing all prior data
 .saveAsTable("workspace.bronze.customers"))  # Save the DataFrame as a managed Unity Catalog table in the workspace.bronze schema
```

```python
# Write the first 10 rows of the customers DataFrame to a managed Unity Catalog table in the workspace.bronze schema

(df                                            # Start from the customers DataFrame loaded in the previous cell
 .limit(num=10)                                # Restrict the DataFrame to the first 10 rows before writing
 .write                                        # Access the DataFrameWriter to persist the DataFrame to a target destination
 .mode("overwrite")                            # Overwrite the target table if it already exists, replacing all prior data
 .saveAsTable("workspace.bronze.customers"))   # Save the limited DataFrame as a managed Unity Catalog table in the workspace.bronze schema
```

```python
# Query all rows and columns from the customers table in the workspace.bronze schema
# using a SQL query via the SparkSession, then display the results in the notebook
df = spark.sql("""
  SELECT *
  FROM workspace.bronze.customers
""")

# Render the DataFrame results as an interactive table in the notebook output
df.display()
```
