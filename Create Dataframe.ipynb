```python
# List of tuples - rows of data
data1 = [('A',2,3), ('B',5,6)]

# create dataframe using spark.createDataFrame method
df = spark.createDataFrame(data=data1)

# display dataframe
df.display()
```

```python
# column names (header)
header = ["name", "age"]

# List of tuples - rows
rows = [("Anil", 32), ("Bhim", 45)]

# create dataframe using spark.createDataFrame method
df = spark.createDataFrame(data=rows, schema=header)

# display dataframe
df.display()
```

```python
from pyspark.sql.types import *

# schema definition
schema = StructType([StructField("name", StringType(), True), \
                     StructField("age", IntegerType(), True)])
# data definition
rows = [("Anil", 32), ("Bhim", 45)]

# create dataframe using spark.createDataFrame method.
df = spark.createDataFrame(data=rows, schema=schema)

# Display the dataframe
df.display()
```
